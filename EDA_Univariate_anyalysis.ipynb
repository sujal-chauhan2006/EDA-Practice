# Univariate EDA (Learning)




##This notebook contains basic univariate analysis performed for learning purposes.


```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as mt
```

```python
df = pd.read_csv("/content/netflix_movies_tv_shows.csv")
```

```python
df.head()  # Give first 5 record as defualt
```

## 1. Categorical Data


```python
df[['type','genre','title']].sample(5)  #This is the categorical column in dataset and then get random 5 record from there
```

### a. countplot

```python
sns.countplot(x = 'genre',data = df)  # Countplot showing distribution of genres
```

### b.pie chart

```python
df['genre'].value_counts().plot(kind='pie', autopct='%.2f')
# This is the pie chart give how much part it occuipied in pie chart
```

## 2. Numeric Data

```python
df[['release_year','imdb_rating','vote_count','budget_million_usd','collection_million_usd','popularity_score','awards_won']]
```

### a.histpgram

```python
mt.hist(df['budget_million_usd'],bins = 40)
# This is histogram give numeric data
# In here it give how many movie has larger budget and has smaller budget for tv show and movies
```

### b. Distplot

```python
sns.distplot(df['popularity_score'],bins = 30)
# This is distplot it give skewness(Normal Distribution)
```

```python
df['popularity_score'].skew()  #In pandas it has built in function for skewness
```

### c. Boxplot

```python
sns.boxplot(df['vote_count'])
# Boxplot shows median, quartiles and outliers
```
