# Exploratory Data Analysis (EDA)



## Bivariate and Multivariate Analysis

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as mt
```

```python
# Some dataset that seaborn provide

tips = sns.load_dataset("tips")
filghts = sns.load_dataset("flights")
iris = sns.load_dataset("iris")
```

```python
print(filghts)
print(iris)
```

```python
tips.head(10)
# First 10 Record from dataset
```

```python
tips.shape
# Give dataset count of rows and column
```

## 1. Scatter plot (Num - Num)

```python
# Bivariate
sns.scatterplot(x=tips['total_bill'],y=tips['tip'],color='black')
# Tip tends to increase as total bill increases, indicating a positive correlation.
```

```python
# Multivariate
sns.scatterplot(x=tips['total_bill'],y=tips['tip'],hue=tips['sex'],style=tips['smoker'],size=tips['size'],palette={'Male':'blue','Female':'black'} )
```

## 2.Bar plot (Num-cate)

```python
print(iris.head())
```

```python
sns.barplot(x=iris['species'],y=iris['sepal_width'],palette=['green','red','blue'])
# This describe which type of flower contain more width
```

## 3. Bax plot (Num - Cate)

```python
filghts.head()
```

```python
sns.boxplot(y = filghts['passengers'],x = filghts['month'])
# This chart describe which month has most passengers
```

## 4. Distplot (Num - Cate)

```python
tips.sample(5)
```

```python
sns.distplot(tips[tips['sex']=='Female']['tip'],hist=False,color='cyan')
sns.distplot(tips[tips['sex']=='Male']['tip'],hist=False,color='red')

# Distplot use to show which gender pay more tip in restruent
```

## 5.Heatmap (cate - cate)

```python
tips.head(3)
```

```python
sns.heatmap(pd.crosstab(tips['sex'],tips['smoker']),cmap='Reds')

# This chart show which gender do more smoke
```

```python
sns.heatmap(filghts.pivot_table(index='year',columns='month',values='passengers'))

# This chart which month has more passenger to travel and which month has less passernger in travel
```

## 6.clustermap (cate - cate)

```python
tips.head(3)
```

```python
sns.clustermap(pd.crosstab(tips['day'],tips['time']),cmap='coolwarm',linewidths=.5,linecolor='Black')

# Clustermap shows Similarity between rows & columns (tree/dendrogram)
```

```python
sns.clustermap(filghts.pivot_table(index='year',columns='month',values='passengers'))

# This give relation between simalrity between multiple months of years
```

## 7.Pairplot

```python
sns.pairplot(iris)

# Pairplot automatically creates visualizations for all numeric columns and shows their relationships with each other.
```

```python
sns.pairplot(iris,hue = 'species')

# This Pairplot give different Species length and width
```
